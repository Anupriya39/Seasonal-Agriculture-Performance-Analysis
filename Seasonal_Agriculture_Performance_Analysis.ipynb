## Univariate: Season distribution (Pie Chart)

```python
plt.figure(figsize=(7, 7))
df['Season'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, cmap='viridis')
plt.title('Distribution of Records by Season')
plt.ylabel('') # Hide the default 'Season' label on the y-axis
plt.tight_layout()
plt.show()
```

```python
# IQR-based outlier summary for numerical columns

outlier_summary = []

for col in numeric_columns:
  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  lower = q1 - 1.5 * iqr
  upper = q3 + 1.5 * iqr
  count = ((df[col] < lower) | (df[col] > upper)).sum()

  outlier_summary.append({
      "Column": col,
      "Lower Bound": lower,
      "Upper_Bound": upper,
      "Outlier_Count": count,
      "Outlier_Percentage": round(count / len(df) * 100, 2)
  })

outlier_summary= pd.DataFrame (outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)
outlier_summary
```

```python
# Student Analysis 1

# Question:
# Which crops perform best in each season based on average yield?

# Analysis:
# Calculate the average yield of each crop in each season.

crop_season_yield = (
    df.groupby(['Season', 'Crop'])['Yield_Tonnes_Ha']
    .mean()
    .reset_index()
)

# Find the best-performing crop in each season
best_crops = crop_season_yield.loc[
    crop_season_yield.groupby('Season')['Yield_Tonnes_Ha'].idxmax()
]

print("Best performing crop in each season:")
print(best_crops)

# Visualization
plt.figure(figsize=(12, 6))

sns.barplot(
    data=crop_season_yield,
    x='Crop',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title('Crop Yield Comparison Across Seasons')
plt.xlabel('Crop')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=45)
plt.show()

# Interpretation:
# The analysis identifies which crops achieve the highest average
# yield in different seasons. This helps understand crop-specific
# seasonal patterns and suitability.

# Relevance:
# This analysis is relevant because identifying suitable crops
# for different seasons can support better agricultural planning
# and crop selection.
```

```python
# Student Analysis 2

# Question:
# Is rainfall associated with agricultural production?

# Analysis:
# Calculate the correlation between rainfall and production.

correlation = df[['Rainfall_mm', 'Production_Tonnes']].corr()

print("Correlation between Rainfall and Production:")
print(correlation)

# Visualization
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='Rainfall_mm',
    y='Production_Tonnes',
    hue='Season'
)

plt.title('Relationship Between Rainfall and Agricultural Production')
plt.xlabel('Rainfall (mm)')
plt.ylabel('Production (Tonnes)')
plt.show()

# Interpretation:
# The scatter plot and correlation value show whether rainfall
# and agricultural production have a positive, negative, or
# weak relationship.

# Relevance:
# Rainfall is an important environmental factor in agriculture.
# Understanding its relationship with production can help identify
# seasonal environmental patterns.

```

```python
# Student Analysis 3

# Question:
# Which season provides the best economic performance based on profit?

# Analysis:
# Calculate the average, median, and standard deviation of profit
# for each season.

season_profit = (
    df.groupby('Season')['Profit_INR']
    .agg(['mean', 'median', 'std'])
    .sort_values('mean', ascending=False)
)

print("Seasonal Profit Analysis:")
print(season_profit)

# Visualization
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x='Season',
    y='Profit_INR'
)

plt.title('Profit Distribution Across Seasons')
plt.xlabel('Season')
plt.ylabel('Profit (INR)')
plt.show()

# Interpretation:
# The analysis identifies the season with the highest average profit
# and shows the variation in profit within each season.

# Relevance:
# Economic performance is an important part of agricultural planning.
# Comparing seasonal profit can help identify financially favorable
# agricultural periods.
```
